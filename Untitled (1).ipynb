```python
import asyncio
import time
from concurrent.futures import ThreadPoolExecutor
from functools import lru_cache
import random
```

```python
async def async_task(task_id):
    await asyncio.sleep(random.uniform(0.1, 0.5))
    return f"Task {task_id} completed"
```

```python
def sync_task(task_id):
    time.sleep(random.uniform(0.1, 0.5))
    return f"Task {task_id} done"

def run_parallel_tasks(n=10):
    with ThreadPoolExecutor(max_workers=5) as executor:
        results = list(executor.map(sync_task, range(n)))
    return results

start = time.time()
results = run_parallel_tasks(10)
end = time.time()

print("Parallel Results:", results)
print(f"Parallel Time Taken: {end - start:.2f} seconds")
```

```python
@lru_cache(maxsize=100)
def expensive_computation(x):
    time.sleep(0.5)
    return x * x

start = time.time()
print(expensive_computation(10))
print(expensive_computation(10))  # Cached
end = time.time()

print(f"Cached Execution Time: {end - start:.2f} seconds")
```

```python
def measure_performance():
    # Without caching
    start = time.time()
    for i in range(5):
        expensive_computation.cache_clear()
        expensive_computation(i)
    no_cache = time.time() - start

    # With caching
    start = time.time()
    for i in range(5):
        expensive_computation(i)
    with_cache = time.time() - start

    print(f"Without Cache: {no_cache:.2f}s")
    print(f"With Cache: {with_cache:.2f}s")

measure_performance()
```

class ProcessingStrategy:
    async def process(self, data):
        pass

class FastStrategy(ProcessingStrategy):
    async def process(self, data):
        await asyncio.sleep(0.1)
        return f"Fast processed {data}"

class SlowStrategy(ProcessingStrategy):
    async def process(self, data):
        await asyncio.sleep(1)
        return f"Slow processed {data}"

class Processor:
    def __init__(self, strategy):
        self.strategy = strategy

    async def execute(self, data):
        return await self.strategy.process(data)

processor = Processor(FastStrategy())
result = asyncio.run(processor.execute("data"))
print(result)
